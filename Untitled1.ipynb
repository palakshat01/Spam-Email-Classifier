```python
"""
==========================================================
 BEGINNER-FRIENDLY SPAM CLASSIFIER
==========================================================
This script builds a simple Machine Learning model that can
tell the difference between "spam" and "ham" (normal) messages.

Steps we follow (matching how real ML projects are built):
  1. Load a dataset (real SMS Spam Collection file if you have
     it, otherwise a small built-in sample dataset).
  2. Clean/preprocess the text.
  3. Turn text into numbers using TF-IDF.
  4. Train a Naive Bayes model.
  5. Evaluate the model (accuracy, precision, recall, confusion matrix).
  6. Let you type your own message and check if it's spam.

Every step has comments explaining WHY we do it, not just what
the code does.
==========================================================
"""

# ---------------------------------------------------------
# STEP 0: IMPORT THE LIBRARIES WE NEED
# ---------------------------------------------------------
import re                     # 're' = regular expressions, used to clean text
import string                 # gives us a list of punctuation characters
import pandas as pd           # pandas lets us work with data in table form (like Excel)
import os

from sklearn.model_selection import train_test_split       # splits data into train/test sets
from sklearn.feature_extraction.text import TfidfVectorizer # turns text into numbers
from sklearn.naive_bayes import MultinomialNB               # the ML model we'll train
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report,
)


# ---------------------------------------------------------
# STEP 1: LOAD THE DATASET
# ---------------------------------------------------------
def load_dataset():
    """
    Tries to load the real SMS Spam Collection dataset if a file
    called 'spam.csv' or 'SMSSpamCollection' exists in the current
    folder. If not found, it falls back to a small hand-made sample
    dataset so the script still runs end-to-end.

    The real dataset (free, public) can be downloaded from:
    https://archive.ics.uci.edu/dataset/228/sms+spam+collection

    It's a tab-separated file with two columns: label ('ham'/'spam')
    and the message text. If you download it, place it next to this
    script as 'SMSSpamCollection' and this function will find it.
    """

    # --- Try to load the real dataset first ---
    if os.path.exists("SMSSpamCollection"):
        # The real file has no header row, columns separated by TAB
        df = pd.read_csv(
            "SMSSpamCollection", sep="\t", header=None, names=["label", "message"]
        )
        print("Loaded real SMS Spam Collection dataset.")
        return df

    if os.path.exists("spam.csv"):
        # Some versions of this dataset are saved as CSV with extra columns
        df = pd.read_csv("spam.csv", encoding="latin-1")
        df = df.iloc[:, :2]                      # keep only first 2 columns
        df.columns = ["label", "message"]        # rename them
        print("Loaded spam.csv dataset.")
        return df

    # --- Fallback: small built-in sample dataset ---
    # This mimics the structure of the real dataset: a label
    # ("spam" or "ham") and a message. It's small on purpose,
    # just enough to demonstrate the full pipeline.
    print("No dataset file found — using a small built-in sample dataset instead.")
    data = {
        "label": [
            "ham", "ham", "spam", "ham", "spam", "ham", "spam", "ham",
            "spam", "ham", "ham", "spam", "ham", "spam", "ham", "spam",
            "ham", "ham", "spam", "ham", "spam", "ham", "ham", "spam",
            "ham", "spam", "ham", "ham", "spam", "ham",
        ],
        "message": [
            "Hey, are we still meeting for lunch tomorrow?",
            "Can you send me the report before 5pm?",
            "WINNER!! You have been selected to receive a $1000 cash prize. Click here to claim now!",
            "Mom, I'll be home late tonight, don't wait up.",
            "URGENT! Your account has been suspended. Verify your details immediately to avoid closure.",
            "Let's catch up this weekend, it's been too long.",
            "Congratulations! You've won a free iPhone. Text WIN to 12345 to claim your prize now.",
            "Don't forget to pick up milk on your way home.",
            "Get cheap loans instantly! No credit check required. Apply now at loanfast.com",
            "The meeting has been moved to 3pm, see you there.",
            "Happy birthday! Hope you have an amazing day.",
            "FREE entry into our weekly draw for a chance to win cash prizes! Text ENTER now.",
            "Can we reschedule our call to Thursday?",
            "Limited time offer! Buy one get one free on all items. Shop now before it's gone!",
            "I finished the assignment, sending it over now.",
            "You have been chosen for a special cash reward. Reply YES to claim your $500 now.",
            "Thanks for your help earlier, really appreciate it.",
            "Are you coming to the gym later today?",
            "Claim your free vacation package now! Limited slots available, call now.",
            "Please review the attached document and let me know your thoughts.",
            "Congratulations, you've been selected for a $5000 Walmart gift card, click to claim.",
            "See you at the party tonight, don't be late.",
            "I'll call you back in 10 minutes, I'm in a meeting.",
            "Your loan has been approved! Get instant cash transferred to your account today.",
            "Can you pick up the kids from school today?",
            "Act now! Your subscription is about to expire, renew today for a huge discount.",
            "Good morning! Hope you slept well.",
            "Let's grab coffee sometime this week.",
            "You've won a lottery prize worth $10,000. Send your bank details to claim it now.",
            "Reminder: your dentist appointment is at 4pm tomorrow.",
        ],
    }
    return pd.DataFrame(data)


df = load_dataset()
print(f"\nDataset shape: {df.shape[0]} rows, {df.shape[1]} columns")
print(df["label"].value_counts())   # shows how many spam vs ham messages we have


# ---------------------------------------------------------
# STEP 2: PREPROCESS THE TEXT
# ---------------------------------------------------------
# A small, common list of English "stopwords" — very frequent
# words (the, is, at, etc.) that usually don't help distinguish
# spam from ham, so we remove them. (We avoid using the nltk
# library here to keep dependencies minimal for beginners.)
STOPWORDS = set("""
a an the and or but if while is are was were be been being to of in on
for with as by at from this that these those it its it's i you he she
we they me him her them my your his her our their so than too very
can could will would should just not no nor do does did doing have
has had having up down out over under again further then once here
there when where why how all any both each few more most other some
such only own same
""".split())


def clean_text(text):
    """
    Cleans a single piece of text so the model can learn from it
    more easily. Steps:
      1. Lowercase everything      -> "FREE Money" and "free money" become the same word
      2. Remove punctuation        -> "!!!" and "$" carry little meaning on their own
      3. Remove numbers            -> optional; here we keep digits out for simplicity
      4. Remove extra whitespace
      5. Remove stopwords          -> removes common filler words
    """
    text = text.lower()                                        # 1. lowercase

    text = re.sub(r"http\S+|www\S+", " ", text)                # remove URLs
    text = text.translate(str.maketrans("", "", string.punctuation))  # 2. remove punctuation
    text = re.sub(r"\d+", " ", text)                           # 3. remove digits
    text = re.sub(r"\s+", " ", text).strip()                   # 4. collapse multiple spaces

    # 5. remove stopwords, word by word
    words = text.split()
    words = [w for w in words if w not in STOPWORDS]

    return " ".join(words)


# Apply the cleaning function to every message in the dataset.
# .apply() runs clean_text() on each row of the "message" column.
df["clean_message"] = df["message"].apply(clean_text)

print("\nExample of cleaning:")
print("BEFORE:", df["message"].iloc[2])
print("AFTER :", df["clean_message"].iloc[2])


# ---------------------------------------------------------
# STEP 3: CONVERT TEXT TO NUMBERS (TF-IDF)
# ---------------------------------------------------------
# Machine Learning models only understand numbers, not raw text.
# TF-IDF (Term Frequency - Inverse Document Frequency) converts
# each message into a vector of numbers where:
#   - words that appear often IN THIS message get a higher score
#   - words that appear in MANY messages (common words) get a lower score
# This helps the model focus on words that are actually distinctive.
vectorizer = TfidfVectorizer()

# Convert the labels into numbers too: spam -> 1, ham -> 0
# This is required because ML models work with numeric labels.
y = df["label"].map({"ham": 0, "spam": 1})
X_text = df["clean_message"]


# ---------------------------------------------------------
# STEP 4: SPLIT DATA & TRAIN THE MODEL
# ---------------------------------------------------------
# We split our data into a "training set" (used to teach the model)
# and a "test set" (used to check how well it learned, on data it
# has never seen before). test_size=0.25 means 25% goes to testing.
X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text, y, test_size=0.25, random_state=42, stratify=y
)

# IMPORTANT: fit_transform() on training data LEARNS the vocabulary
# and converts training text to numbers.
# transform() on test data reuses that SAME vocabulary — we never
# let the model "see" test data while learning, to avoid cheating.
X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

# Create and train the Naive Bayes model.
# MultinomialNB is a classic, simple, and effective algorithm for
# text classification tasks like spam detection.
model = MultinomialNB()
model.fit(X_train, y_train)   # this is where the actual "learning" happens

print("\nModel trained successfully!")


# ---------------------------------------------------------
# STEP 5: EVALUATE THE MODEL
# ---------------------------------------------------------
# Use the trained model to predict labels for the test set
# (data the model has never seen before).
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
cm = confusion_matrix(y_test, y_pred)

print("\n--- MODEL EVALUATION ---")
print(f"Accuracy : {accuracy:.2f}  (fraction of correct predictions overall)")
print(f"Precision: {precision:.2f}  (of messages predicted spam, fraction that really are)")
print(f"Recall   : {recall:.2f}  (of all real spam messages, fraction we caught)")

print("\nConfusion Matrix:")
print("                Predicted Ham   Predicted Spam")
print(f"Actual Ham           {cm[0][0]:<15}{cm[0][1]}")
print(f"Actual Spam          {cm[1][0]:<15}{cm[1][1]}")

print("\nFull classification report:")
print(classification_report(y_test, y_pred, target_names=["ham", "spam"], zero_division=0))


# ---------------------------------------------------------
# STEP 6: TRY YOUR OWN MESSAGE
# ---------------------------------------------------------
def predict_message(message):
    """
    Takes any message you type in and tells you whether the
    trained model thinks it's SPAM or HAM (normal).

    Steps:
      1. Clean the text the same way we cleaned training data.
      2. Convert it to numbers using the SAME vectorizer
         (so the numbers "mean" the same thing as during training).
      3. Ask the model to predict.
      4. Also show the model's confidence (probability).
    """
    cleaned = clean_text(message)                 # step 1
    vectorized = vectorizer.transform([cleaned])  # step 2 (note the list [ ])
    prediction = model.predict(vectorized)[0]     # step 3: 0 = ham, 1 = spam
    probability = model.predict_proba(vectorized)[0]  # [prob_ham, prob_spam]

    label = "SPAM" if prediction == 1 else "HAM (not spam)"
    confidence = probability[prediction] * 100

    return label, confidence


if __name__ == "__main__":
    # A couple of built-in examples so you can see it work immediately
    test_messages = [
        "Congratulations! You have won a free cruise. Call now to claim your prize!",
        "Hey, are you free to grab dinner tonight?",
    ]

    print("\n--- TRYING SOME EXAMPLE MESSAGES ---")
    for msg in test_messages:
        label, confidence = predict_message(msg)
        print(f'Message: "{msg}"')
        print(f"  -> Prediction: {label}  (confidence: {confidence:.1f}%)\n")

    # Interactive mode: type your own message and check it.
    # This loop keeps asking until you type 'quit'.
    print("--- TYPE YOUR OWN MESSAGE (type 'quit' to exit) ---")
    while True:
        user_input = input("\nEnter a message to check: ")
        if user_input.strip().lower() == "quit":
            print("Goodbye!")
            break
        label, confidence = predict_message(user_input)
        print(f"Prediction: {label}  (confidence: {confidence:.1f}%)")
```
